```python
import os
import json
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path
from PIL import Image

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")

print("Path to dataset files:", path)
```

```python
splits=[]
types=[]
print("Dataset path:", path)
print("\nDataset structure:")

for item in os.listdir(path):
    item_path = os.path.join(path, item)

    if os.path.isdir(item_path):
        print(f"Folder:  {item}/")
    else:
        print(f"File {item}")

# print(os.listdir(path))
for ele in os.listdir(path + "/chest_xray"):
  if ele =="chest_xray" or ele =="__MACOSX":
    continue
  splits.append(ele)

for split in splits:
  t=os.listdir(path + "/chest_xray/" + split)
  for ele in t:
    if ele in types:
      continue
    types.append(ele)
print(splits)
print(types)
```

```python
root = Path(path)
images=[]
IMG_SIZE = (180, 180)
for split in splits:
    split_path = root / "chest_xray" / split

    if split_path.exists():
        print(f"\n{split}")

        for class_name in types:
            class_path = split_path / class_name

            if class_path.exists():
              for f in class_path.iterdir():
                if f.is_file():
                  images.append(f)


            print(f"{class_name}: {len(images)} images")


processed_images = []

for image_path in images:
    try:
        with Image.open(image_path) as image:
            image = image.convert("RGB")
            image = image.resize(IMG_SIZE)

            processed_images.append(image.copy())

    except Exception as e:
        print(f"Erreur avec {image_path}: {e}")

print(f"Images traitées avec succès : {len(processed_images)}")
```

```python
train_dir = Path(path) / "chest_xray" / "train"

fig, axes = plt.subplots(2, 4, figsize=(12, 6))

for row, class_name in enumerate(types):
    images = list((train_dir / class_name).glob("*.jpeg"))[:4]
    print(f"\n{class_name}: {len(images)} images")
    for col, img_path in enumerate(images):
        axes[row, col].imshow(Image.open(img_path), cmap="gray")
        axes[row, col].set_title(class_name)
        axes[row, col].axis("off")

plt.show()
```

```python

class_weight_dict = {
    0: total / (2 * counts["NORMAL"]),
    1: total / (2 * counts["PNEUMONIA"]),
}

print("Class weights:", class_weight_dict)
```

```python
train_dir = Path(path) / "chest_xray" / "train"
test_dir = Path(path) / "chest_xray" / "test"

CLASS_NAMES=types
SEED=42

common = dict(
    class_names=CLASS_NAMES,
    label_mode="binary",
    image_size=IMG_SIZE,
    batch_size=32,
)

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir, validation_split=0.2,
    subset="training",
    seed=SEED,
    **common)

val_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    **common)

test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    shuffle=False,
    **common)

print("Classes:", train_ds.class_names)
```

```python
from pathlib import Path
from PIL import Image

valid_images = []
corrupted_images = []

for split in splits:
    for class_name in types:
        class_path = root / split / class_name

        if not class_path.exists():
            print(f"Dossier introuvable : {class_path}")
            continue

        for img_path in class_path.iterdir():
            if not img_path.is_file():
                continue

            try:
                with Image.open(img_path) as img:
                    img.verify()

                valid_images.append(img_path)

            except Exception as e:
                corrupted_images.append(img_path)
                print(f"Image invalide : {img_path} | Erreur : {e}")

print("Images valides :", len(valid_images))
print("Images corrompues :", len(corrupted_images))
```

```python
# Normalisation des pixels entre 0 et 1
normalization_layer = tf.keras.layers.Rescaling(1.0 / 255)

train_ds = train_ds.map(
    lambda images, labels: (normalization_layer(images), labels)
)

val_ds = val_ds.map(
    lambda images, labels: (normalization_layer(images), labels)
)

test_ds = test_ds.map(
    lambda images, labels: (normalization_layer(images), labels)
)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=AUTOTUNE)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
# 13. Construire le modèle CNN
model = models.Sequential([
    # Input
    layers.Input(shape=(180, 180, 3)),

    # Block 1
    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    # Block 3
    layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    # Classification
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),

    # Binary output
    layers.Dense(1, activation="sigmoid")
])

model.summary()
```

```python
import tensorflow as tf
# 14. Compiler et entraîner le modèle
# Compile the model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)
```

```python
# Train the model
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ModelCheckpoint(
        "best_chestscan_cnn.keras",
        monitor="val_loss",
        save_best_only=True
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=callbacks
)
```
